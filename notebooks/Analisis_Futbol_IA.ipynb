```python
#@title 📱 RADAR EN VIVO — toca ▶️ y copia el link
import os, subprocess, sys, time
from IPython.display import HTML, display, clear_output

print("⏳ Preparando el cerebro (2-3 minutos)...")
if not os.path.exists("/content/analisis-futbol-ia"):
    subprocess.run(["git", "clone", "-q", "https://github.com/j2916artprepa2a-ux/analisis-futbol-ia.git", "/content/analisis-futbol-ia"], check=True)
else:
    subprocess.run(["git", "-C", "/content/analisis-futbol-ia", "pull", "-q"])
os.chdir("/content/analisis-futbol-ia")
sys.path.insert(0, "/content/analisis-futbol-ia")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
print("⏳ Cargando la IA...")
subprocess.run([sys.executable, "-m", "futbol_ia.modelos"], check=True, capture_output=True)

import torch
if not torch.cuda.is_available():
    display(HTML("<h2 style='color:#d33'>⚠️ No hay GPU. Ve a Entorno de ejecución → Cambiar tipo de entorno → T4 GPU, y vuelve a tocar ▶️</h2>"))

import servidor_en_vivo as srv
srv.iniciar_en_segundo_plano()
print("⏳ Abriendo el link público...")
URL = srv.abrir_tunel()
clear_output()
display(HTML(f'''
<div style="text-align:center;padding:20px;font-family:sans-serif">
  <h2>✅ ¡Cerebro listo!</h2>
  <p style="font-size:15px">Este es tu link:</p>
  <p style="font-size:17px;word-break:break-all;background:#eef;padding:10px;border-radius:8px;color:#123">{URL}</p>
  <button onclick="navigator.clipboard.writeText('{URL}');this.innerText='✅ COPIADO'"
     style="background:#16a34a;color:white;font-size:24px;padding:16px 34px;border:none;border-radius:14px;font-weight:bold">
     📋 COPIAR LINK
  </button>
  <p style="margin-top:16px;color:#666">Pégalo en la app Radar Fútbol. No cierres esta página.</p>
</div>'''))
while True:
    time.sleep(60)
```
